```python
# Time Analysis
df["InvoiceDate"]=pd.to_datetime(df["InvoiceDate"],dayfirst=True,errors="coerce")
df["month"]=df["InvoiceDate"].dt.to_period("M")
```

```python
monthly_rev=df.groupby("month")["Revenue"].sum()
print(monthly_rev)
```

```python
best_month=monthly_rev.idxmax()
best_rev=monthly_rev.max()

worst_month=monthly_rev.idxmin()
worst_rev=monthly_rev.min()

print("Best Month:",best_month,best_rev)
print("\n","Worst Month:",worst_month,worst_rev)
```

```python
monthly_Quantity=df[df["Quantity"]>0].groupby("month")["Quantity"].sum()
print(monthly_Quantity)
```

```python
print("Highest quantity month:",monthly_Quantity.idxmax())
print("Lowest quantity month:",monthly_Quantity.idxmin())

print("\n","Highest Quantity:",monthly_Quantity.max())
print("Lowest Quantity:",monthly_Quantity.min())
```

```python
monthly_transactions=df[df["Quantity"]>0].groupby("month")["Invoice"].nunique()
print(monthly_transactions)
print("Highest Transactions month:-",monthly_transactions.idxmax(),"Highest Transactions:-",monthly_transactions.max())
print("\n","Lowest Transactions month:-",monthly_transactions.idxmin(),"Lowest Transactions:-",monthly_transactions.min())
```
